```python
import numpy as np

print("Numpy imported successfully")
```

```python
arr1 = np.array([10,20,30,40,50])
print("1D array:")
print(arr1)
print()
```

```python
arr2 = np.array([
    [1,2,3],
    [4,5,6]
])
print("2D Array:")
print(arr2)
print()
```

```python
print("1D array Dimensions:",arr1.ndim)
print("1D array Shape:",arr1.shape)
print("1D array Size:",arr1.size)
print()
print("2D array Dimensions:",arr2.ndim)
print("2D array Shape:",arr2.shape)
print("2D array Size:",arr2.size)
print()
```

```python
print("first number",arr1[0])
print("second number",arr1[1])
print("third number",arr1[2])
print("fourth number",arr1[3])
print("fifth number",arr1[4])
print()
print("first number",arr2[0,0])
print("second number",arr2[0,1])
print("third number",arr2[0,2])
print()
```

```python
print("array slice[1:4]:")
print(arr1[1:4])
print()
print("element at row 1 column 2:")
print(arr2[0:1])
print()
```

```python
zeros_array = np.zeros((3,3))
print("zeros_array")
print()
ones_array = np.ones((2,4))
print("Ones array")
print(ones_array)
print()
range_array = np.arange(1,11)
print("Range array")
print(range_array)
print()
```
